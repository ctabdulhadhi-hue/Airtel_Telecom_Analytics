```python
q("""SELECT
    ROUND(AVG(stated_arpu), 2) AS average_stated_arpu,
    ROUND(AVG(actual_arpu_6m), 2) AS average_actual_arpu,
    ROUND(AVG(stated_arpu - actual_arpu_6m), 2) AS average_gap,
    ROUND(
        AVG(stated_arpu - actual_arpu_6m) * 100.0
        / NULLIF(AVG(actual_arpu_6m), 0),
        2
    ) AS overstatement_percentage
FROM customer_features;""")
```

```python
q("""SELECT
    category,
    COUNT(*) AS ticket_count,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS share_percentage
FROM complaints
GROUP BY category
ORDER BY ticket_count DESC;""")
```

```python
q("""SELECT
    channel,
    COUNT(*) AS ticket_count,
    ROUND(AVG(is_resolved) * 100, 2) AS resolved_share_percentage
FROM complaints
GROUP BY channel
ORDER BY resolved_share_percentage DESC;""")
```

```python
q("""SELECT
    is_resolved,
    COUNT(*) AS ticket_count,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS perc
FROM complaints
GROUP BY is_resolved
ORDER BY is_resolved;""")
```

```python
q("""
SELECT
    category,
    ROUND(AVG(is_resolved) * 100, 2) AS resolved_share_percentage,
    ROUND(AVG(resolution_days), 2) AS avg_resolution_days
FROM complaints
GROUP BY category
ORDER BY resolved_share_percentage ASC, avg_resolution_days DESC;
""")
```

```python
q("""
SELECT
    AVG(resolution_days) AS avg_resolution_days,
    MIN(resolution_days) AS min_resolution_days,
    MAX(resolution_days) AS max_resolution_days
FROM complaints
WHERE is_resolved = 1;
""")
```

```python
q("""
SELECT
    complaint_count,
    COUNT(*) AS subscriber_count
FROM customer_features
GROUP BY complaint_count
ORDER BY complaint_count;
""")
```

```python
q("""
SELECT
    cf.circle_name,
    ROUND(COUNT(c.complaint_id) * 1.0 / COUNT(DISTINCT cf.customer_id), 4) AS complaints_per_subscriber
FROM customer_features cf
LEFT JOIN complaints c
    ON cf.customer_id = c.customer_id
GROUP BY cf.circle_name
ORDER BY complaints_per_subscriber DESC
LIMIT 5;
""")
```
