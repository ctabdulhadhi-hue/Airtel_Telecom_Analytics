```python
airtel_con = sqlite3.connect('../data/raw/airtel.db')

pd.read_sql_query("""
WITH l AS (
    SELECT customer_id,
           MAX(usage_date) AS last_use
    FROM usage_daily
    GROUP BY customer_id
)
SELECT c.is_churned,
       COUNT(*) AS subs,
       ROUND(AVG(julianday('2025-06-30') - julianday(l.last_use)), 2) AS avg_quiet,
       MIN(julianday('2025-06-30') - julianday(l.last_use)) AS min_,
       MAX(julianday('2025-06-30') - julianday(l.last_use)) AS max_
FROM customers c
JOIN l ON l.customer_id = c.customer_id
GROUP BY c.is_churned;
""", airtel_con)
```

```python
# Query each available Airtel table to avoid failing on tables that do not exist
airtel_tables = pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
""", airtel_con)['name'].tolist()

date_queries = [
    ('usage_daily.usage_date', 'usage_daily', 'usage_date'),
    ('customers.activation_date', 'customers', 'activation_date'),
    ('customers.churn_date', 'customers', 'churn_date'),
    ('recharges.recharge_date', 'recharges', 'recharge_date'),
]

queries = []
for label, table, date_column in date_queries:
    if table in airtel_tables:
        queries.append(
            f"""SELECT '{label}' AS col,
                       MIN({date_column}) AS first_,
                       MAX({date_column}) AS last_
                FROM {table}"""
        )

pd.read_sql_query(" UNION ALL ".join(queries) + ";", airtel_con)
```
